```python
import pandas as pd

BEHAVIOR_COLUMNS = [
    "impression_id",
    "user_id",
    "time",
    "history",
    "impressions"
]

behaviors = pd.read_csv(
    "../data/raw/valid/behaviors.tsv",
    sep="\t",
    names=BEHAVIOR_COLUMNS
)

behaviors["time"] = pd.to_datetime(
    behaviors["time"],
    format="%m/%d/%Y %I:%M:%S %p"
)

user_counts = behaviors.groupby("user_id").size()

print(user_counts.describe())

print("\nUsers with more than one behavior:")
print((user_counts > 1).sum())

print("\nMaximum behaviors for one user:")
print(user_counts.max())
```

```python
user_counts = behaviors.groupby("user_id").size()

user_id = user_counts.idxmax()

user_data = (
    behaviors[behaviors["user_id"] == user_id]
    .sort_values("time")
)

print("User:", user_id)
print("Number of events:", len(user_data))

user_data[
    ["time", "history", "impressions"]
]
```

```python
for i, row in user_data.iterrows():
    history = row["history"].split() if pd.notna(row["history"]) else []

    clicked_now = [
        item.split("-")[0]
        for item in row["impressions"].split()
        if item.endswith("-1")
    ]

    print(
        row["time"],
        "| history length:", len(history),
        "| clicked now:", clicked_now
    )
```

```python
rows = user_data.reset_index(drop=True)

for i in range(1, len(rows)):
    prev_clicks = {
        item.split("-")[0]
        for item in rows.iloc[i - 1]["impressions"].split()
        if item.endswith("-1")
    }

    current_history = set(
        rows.iloc[i]["history"].split()
    )

    print(
        rows.iloc[i]["time"],
        "| previous clicks now in history:",
        prev_clicks & current_history
    )
```

```python
rows = user_data.reset_index(drop=True)

for i in range(1, len(rows)):
    previous_history = set(rows.iloc[i - 1]["history"].split())
    current_history = set(rows.iloc[i]["history"].split())

    added = current_history - previous_history
    removed = previous_history - current_history

    print(
        rows.iloc[i]["time"],
        "| added:", len(added),
        "| removed:", len(removed)
    )
```
